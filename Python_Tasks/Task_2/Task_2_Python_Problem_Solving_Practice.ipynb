**Q1. Find the Shortest Word in a Sentence**

Write a Python program to find the shortest word in a given sentence.

**Example input:**
Python makes coding easy

**Expected output:**
easy

```python
def shortest_word (sentence):
  words = sentence.split()
  shortest = words[0]
  for word in words:
    if len(word) < len(shortest):
      shortest = word
  return shortest

print(shortest_word("Python makes coding easy"))
print(shortest_word("I am learning Python"))
```

**Q2. Find the Largest Number in a List**

Write a Python program to find the largest number in a list without using the built-in max() function.

**Example input:**
[12, 45, 7, 89, 23]

**Expected output:**
89


```python
def find_largest(numbers):
    if not numbers:
        return "List is empty"
    largest = numbers[0]
    for num in numbers:
        if num > largest:
            largest = num
    return largest

print(find_largest([12, 45, 7, 89, 23]))
print(find_largest([-5, -2, -9, -1]))
```

**Q6. Count Vowels in a String**

Write a Python program to count the total number of vowels in a string. Consider both uppercase and lowercase vowels.

**Example input:**
Education

**Expected output:**
5


```python
def count_vowels(text):
    count = 0
    for ch in text:
        if ch.lower() in "aeiou":
            count += 1
    return count

print(count_vowels("Education"))
print(count_vowels("RHYTHM"))
```

**Q7. Count Consonants in a String**

Write a Python program to count the consonants in a string. Ignore spaces, digits, and special characters.

**Example input:**
Python 123!

**Expected output:**
5


```python
def count_consonants(text):
    count = 0
    for ch in text:
        if ch.isalpha() and ch.lower() not in "aeiou":
            count += 1
    return count

print(count_consonants("Python 123!"))
print(count_consonants("Hello World"))
```

**Q8. Find the Most Frequent Character**

Write a Python program to find the character that appears most frequently in a string. If multiple characters have the same highest frequency, return the first one encountered. Ignore spaces.

**Example input:**
banana

**Expected output:**
a


```python
def most_frequent_char(text):
    counts = {}
    for ch in text:
        if ch != " ":
            counts[ch] = counts.get(ch, 0) + 1
    if not counts:
        return "No characters found"
    best_char = None
    best_count = 0
    for ch, c in counts.items():
        if c > best_count:
            best_count = c
            best_char = ch
    return best_char

print(most_frequent_char("banana"))
print(most_frequent_char("abcabc"))
```

Q10. Find All Characters Appearing More Than Once

Write a Python program to identify all characters that appear more than once in a string. Display each repeated character only once, in the order of its first appearance. Ignore spaces.

Example input:
programming

Expected output:
['r', 'g', 'm']


```python
def repeated_chars(text):
    counts = {}
    for ch in text:
        if ch != " ":
            counts[ch] = counts.get(ch, 0) + 1
    result = []
    for ch in text:
        if ch != " " and counts[ch] > 1 and ch not in result:
            result.append(ch)
    return result

print(repeated_chars("programming"))
print(repeated_chars("hello world"))
```

Q15. Create a Dictionary of Numbers and Their Even/Odd Status

Write a Python program to create a dictionary containing each number as a key and "Even" or "Odd" as its value.

Example input:
[1, 2, 3, 4, 5]

Expected output:
{1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even', 5: 'Odd'}


```python
def even_odd_dict(numbers):
    result = {}
    for num in numbers:
        if num % 2 == 0:
            result[num] = "Even"
        else:
            result[num] = "Odd"
    return result

print(even_odd_dict([1, 2, 3, 4, 5]))
print(even_odd_dict([0, -1, -2, 10]))
```

Q18. Find the First Word That Appears Only Once

Write a Python program to find the first word that occurs exactly once in a sentence. Treat uppercase and lowercase versions as the same word.

Example input:
apple banana apple mango banana orange

Expected output:
mango


```python
def first_unique_word(sentence):
    words = sentence.lower().split()
    counts = {}
    for word in words:
        counts[word] = counts.get(word, 0) + 1
    for word in words:
        if counts[word] == 1:
            return word
    return "No unique word found"

print(first_unique_word("apple banana apple mango banana orange"))
print(first_unique_word("Dog cat DOG bird"))
```

Q20. Find the Longest Word That Appears Only Once

Write a Python program to find the longest word that appears exactly once in a sentence. If multiple qualifying words have the same maximum length, return the first one encountered. Treat uppercase and lowercase versions as the same word.

Example input:
Python makes coding useful Python

Expected output:
coding


```python
def longest_unique_word(sentence):
    words = sentence.lower().split()
    counts = {}
    for word in words:
        counts[word] = counts.get(word, 0) + 1
    longest = ""
    for word in words:
        if counts[word] == 1 and len(word) > len(longest):
            longest = word
    if longest == "":
        return "No unique word found"
    return longest

print(longest_unique_word("Python makes coding useful Python"))
print(longest_unique_word("sun moon sun stars moon"))
```
