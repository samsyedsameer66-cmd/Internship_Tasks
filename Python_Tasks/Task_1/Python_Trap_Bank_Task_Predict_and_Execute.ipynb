```python
#Q45
def test(*args):
    print(type(args))

test("a", "b", "c")

# *args always stores the collected positional arguments in a tuple, regardless of their values.
```

```python
#Q46
def test(**kwargs):
    print(kwargs)

test(name="Ram", age=25)
```
