## Q1. Find the Frequency of Each Word in a Sentence

**Problem:** Count how many times each word appears in a sentence and store it in a dictionary.

**Pseudocode:**

1. Create an empty dictionary
2. Split the sentence into words
3. For each word, add 1 to its count (start from 0 if new)
4. Return the dictionary

**Expected output (given example):** `{'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}`

```python
def word_frequency(sentence):
    freq = {}
    for word in sentence.split():
        freq[word] = freq.get(word, 0) + 1
    return freq

# Tests
print(word_frequency("python is easy and python is powerful"))  # given example
print(word_frequency("to be or not to be"))  # extra test -> {'to': 2, 'be': 2, 'or': 1, 'not': 1}
```

## Q2. Find the Frequency of Each Character in a Word

**Problem:** Count how many times each character appears in a word and store it in a dictionary.

**Pseudocode:**

1. Create an empty dictionary
2. Go through each character of the word
3. Add 1 to that character's count
4. Return the dictionary

**Expected output (given example):** `{'b': 1, 'a': 3, 'n': 2}`

```python
def char_frequency(word):
    freq = {}
    for ch in word:
        freq[ch] = freq.get(ch, 0) + 1
    return freq

# Tests
print(char_frequency("banana"))  # given example
print(char_frequency("hello"))  # extra test -> {'h': 1, 'e': 1, 'l': 2, 'o': 1}
```

## Q3. Find the First Non-Repeating Character in a Word

**Problem:** Find the first character that appears only once. Return None if there is none.

**Pseudocode:**

1. Count every character
2. Go through the word from left to right
3. Return the first character whose count is 1
4. If none found, return None

**Expected output (given example):** `w`

```python
def first_non_repeating(word):
    freq = char_frequency(word)
    for ch in word:
        if freq[ch] == 1:
            return ch
    return None

# Tests
print(first_non_repeating("swiss"))  # given example
print(first_non_repeating("aabb"))  # extra test -> None
```

## Q4. Check Whether Two Strings Are Anagrams

**Problem:** Check if two strings have the same characters with the same frequencies.

**Pseudocode:**

1. Count characters of the first string
2. Count characters of the second string
3. If both counts are equal, return True, else False

**Expected output (given example):** `True`

```python
def are_anagrams(s1, s2):
    return char_frequency(s1) == char_frequency(s2)

# Tests
print(are_anagrams("listen", "silent"))  # given example
print(are_anagrams("hello", "world"))  # extra test -> False
```

## Q5. Find the First Repeating Character in a String

**Problem:** Find the first character whose second occurrence is reached while reading left to right.

**Pseudocode:**

1. Make an empty set called seen
2. Read each character from left to right
3. If it is already in seen, return it
4. Otherwise add it to seen
5. If the loop ends, return None

**Expected output (given example):** `r`

```python
def first_repeating(text):
    seen = set()
    for ch in text:
        if ch in seen:
            return ch
        seen.add(ch)
    return None

# Tests
print(first_repeating("programming"))  # given example
print(first_repeating("abc"))  # extra test -> None
```

## Q6. Move All Zeros to the End of a List

**Problem:** Move all zeros to the end while keeping the order of the other numbers.

**Pseudocode:**

1. Collect all non-zero numbers in order
2. Count how many zeros were removed
3. Add that many zeros at the end
4. Return the new list

**Expected output (given example):** `[3, 1, 2, 4, 0, 0, 0]`

```python
def move_zeros(nums):
    non_zeros = [n for n in nums if n != 0]
    zeros = [0] * (len(nums) - len(non_zeros))
    return non_zeros + zeros

# Tests
print(move_zeros([0, 3, 0, 1, 2, 0, 4]))  # given example
print(move_zeros([0, 0, 5]))  # extra test -> [5, 0, 0]
```

## Q7. Find the Second Highest Distinct Number in a List

**Problem:** Find the second-largest distinct number. Return None if it does not exist.

**Pseudocode:**

1. Keep two variables: first (largest) and second
2. For each number: if it beats first, move first into second and update first
3. Else if it is different from first and beats second, update second
4. Return second (None if never set)

**Expected output (given example):** `30`

```python
def second_highest(nums):
    first = None
    second = None
    for n in nums:
        if first is None or n > first:
            if first is not None:
                second = first
            first = n
        elif n != first and (second is None or n > second):
            second = n
    return second

# Tests
print(second_highest([10, 20, 40, 30, 40, 20]))  # given example
print(second_highest([5, 5, 5]))  # extra test -> None
```

## Q8. Remove Duplicates Without Changing the Original Order

**Problem:** Remove duplicates but keep the order of first occurrences.

**Pseudocode:**

1. Make an empty set (seen) and an empty list (result)
2. For each number: if not in seen, add to seen and to result
3. Return result

**Expected output (given example):** `[4, 2, 5, 7]`

```python
def remove_duplicates(nums):
    seen = set()
    result = []
    for n in nums:
        if n not in seen:
            seen.add(n)
            result.append(n)
    return result

# Tests
print(remove_duplicates([4, 2, 4, 5, 2, 7, 5]))  # given example
print(remove_duplicates([1, 1, 1]))  # extra test -> [1]
```

## Q9. Find the Missing Number from 1 to N

**Problem:** Find the one missing number in a list of numbers from 1 to N.

**Pseudocode:**

1. Calculate the expected sum of 1 to N using N*(N+1)/2
2. Add up the numbers in the list
3. Missing number = expected sum - actual sum

**Expected output (given example):** `3`

```python
def find_missing(nums, n):
    expected_sum = n * (n + 1) // 2
    return expected_sum - sum(nums)

# Tests
print(find_missing([1, 2, 4, 5, 6], 6))  # given example
print(find_missing([2, 3, 4, 5], 5))  # extra test -> 1
```

## Q10. Find Two Numbers Whose Sum Equals a Target

**Problem:** Find the pair whose second element appears earliest. Return [] if none.

**Pseudocode:**

1. Move j from left to right through the list
2. For each j, check every earlier element i
3. If nums[i] + nums[j] equals target, return [nums[i], nums[j]]
4. If nothing found, return []

**Expected output (given example):** `[2, 7]`

```python
def two_sum_pair(nums, target):
    for j in range(len(nums)):
        for i in range(j):
            if nums[i] + nums[j] == target:
                return [nums[i], nums[j]]
    return []

# Tests
print(two_sum_pair([2, 7, 11, 15], 9))  # given example
print(two_sum_pair([1, 2, 3], 10))  # extra test -> []
```

## Q11. Find the Number Closest to Zero

**Problem:** Find the number closest to zero. If tied, return the positive number.

**Pseudocode:**

1. Assume the first number is the best
2. For each number compare its absolute value with the best
3. If smaller, or equal and the number is bigger (positive), update best
4. Return best

**Expected output (given example):** `2`

```python
def closest_to_zero(nums):
    if not nums:
        return None
    best = nums[0]
    for n in nums:
        if abs(n) < abs(best) or (abs(n) == abs(best) and n > best):
            best = n
    return best

# Tests
print(closest_to_zero([-8, -3, 2, 5, -2]))  # given example
print(closest_to_zero([-4, 4, 9]))  # extra test -> 4
```

## Q12. Find Common Elements Between Two Lists

**Problem:** Common elements, each once, in the order of the first list.

**Pseudocode:**

1. Turn list2 into a set for fast lookup
2. Go through list1 in order
3. If the item is in list2 and not already in result, add it
4. Return result

**Expected output (given example):** `[3, 4, 5]`

```python
def common_elements(list1, list2):
    in_second = set(list2)
    result = []
    for n in list1:
        if n in in_second and n not in result:
            result.append(n)
    return result

# Tests
print(common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))  # given example
print(common_elements([1, 1, 2], [2, 1]))  # extra test -> [1, 2]
```

## Q13. Rotate a List to the Right by K Positions

**Problem:** Rotate a list to the right by K positions.

**Pseudocode:**

1. If the list is empty, return []
2. Reduce K using K % length (so large K still works)
3. Last K elements go to the front, the rest follow
4. Return the new list

**Expected output (given example):** `[4, 5, 1, 2, 3]`

```python
def rotate_right(nums, k):
    if not nums:
        return []
    k = k % len(nums)
    if k == 0:
        return nums[:]
    return nums[-k:] + nums[:-k]

# Tests
print(rotate_right([1, 2, 3, 4, 5], 2))  # given example
print(rotate_right([1, 2, 3], 5))  # extra test -> [2, 3, 1]
```

## Q14. Find the Maximum Sum of Two Consecutive Elements

**Problem:** Find the maximum sum of any two neighbouring elements.

**Pseudocode:**

1. Start best = first + second element
2. Loop through each neighbouring pair
3. If the pair's sum is bigger than best, update best
4. Return best

**Expected output (given example):** `14`

```python
def max_consecutive_sum(nums):
    best = nums[0] + nums[1]
    for i in range(1, len(nums) - 1):
        total = nums[i] + nums[i + 1]
        if total > best:
            best = total
    return best

# Tests
print(max_consecutive_sum([4, 7, 2, 9, 5]))  # given example
print(max_consecutive_sum([-3, -1, -5]))  # extra test -> -4
```

## Q15. Find All Pairs of Numbers Whose Sum Equals a Target

**Problem:** Find all unique pairs (two different positions) that add up to the target, sorted.

**Pseudocode:**

1. Make an empty set for pairs
2. Check every pair of positions i < j
3. If the sum equals target, store (smaller, larger) in the set
4. Sort the set and convert each pair to a list

**Expected output (given example):** `[[1, 8], [2, 7], [4, 5]]`

```python
def all_pairs(nums, target):
    pairs = set()
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                pairs.add((min(nums[i], nums[j]), max(nums[i], nums[j])))
    return [list(p) for p in sorted(pairs)]

# Tests
print(all_pairs([2, 4, 3, 5, 7, 8, 1], 9))  # given example
print(all_pairs([1, 1, 2, 2], 3))  # extra test -> [[1, 2]]
```
